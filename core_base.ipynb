```python
import pandas as pd

df = pd.read_csv("butun-nov-trvz-ehtiyatlar-dt_cd_019.csv",delimiter=",")
df
```

```python
df = df.drop(columns=['Unnamed: 1'])
df
```

```python
#Line chart
import matplotlib.pyplot as plt
plt.style.use("default")
plt.figure(figsize=(6.4, 4.8))
plt.grid(True)
plt.title("Production and Import Over Years")
plt.plot(df["Year"], df["İdxal (ton)"]/1000000, marker='o', color='red', label="Imported")
plt.plot(df["Year"], df["İstehsal (ton)"]/1000000, marker='o', color='blue', label="Produced")
plt.xlabel("Years")
plt.xlim(df["Year"].min(), df["Year"].max())
plt.ylabel("Million Tons")
# plt.ylim(df["Tons"].min(), df["Tons"].max())
plt.legend(loc='upper left')
```
